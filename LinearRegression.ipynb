```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#load the datasets
housing = fetch_california_housing(as_frame = True ) #s a Pandas DataFrame so it can be viewed and manipulated as a labeled table.
data = housing.frame
print(data.head())

#value of x and y
x = data[['MedInc']]
y = data['MedHouseVal']

#test_train_split
x_train,x_test,y_train,y_test = train_test_split(
    x,y,
    test_size=0.2,
    random_state=42
)

#Regression & Predication
model = LinearRegression()

model.fit(x_train,y_train)   #getting value of m and c

y_pred = model.predict(x_test)


mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)

r2 = r2_score(y_test,y_pred)

print(f"Mean Squred Error:{mse}")
print(f"Root MSE:{rmse}")
print(f"R2_Score:{r2}")


#Regression Line
plt.scatter(x_train,y_train,color = 'blue',label = 'Actual Data')
plt.plot(x_test,y_pred,color = 'red',label = 'Regression Line')

plt.xlabel('MedInc')
plt.ylabel('MedHouseVal')
plt.title('Linear Regression')
plt.legend()
plt.grid(True)
plt.show()

#Residual Plot
residuals = y_test - y_pred
plt.scatter(y_pred,residuals,color = 'blue')
plt.axhline(y = 0,color = 'red',linestyle = '--')
plt.xlabel('Predicted Values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.grid(True)
plt.show()
```

```python
# 1.Import Libararies
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_california_housing
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score

#2.Load the dataset
housing = fetch_california_housing(as_frame = True)
data = housing.frame
print(data.head())

#3.select the input and target variable
x = data[['MedInc']]
y = data['MedHouseVal']

#4.split to train and test dataset
x_train,x_test,y_train,y_test = train_test_split(
    x,y,
    test_size = 0.2,
    random_state=42
)

#5.Load the model
model = LinearRegression()
model.fit(x_train,y_train)

y_pred = model.predict(x_test)

#6.Evaluate the model
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)                             #Plz check here
r2 = r2_score(y_test,y_pred)

print(f"Mean Squred Error:{mse}")
print(f"Root MSE: {rmse}")
print(f"r2_score:{r2}")


#7.Plotting a Graph
plt.scatter(x_train,y_train,color = 'Blue',label = "Actual Data")
plt.plot(x_test,y_pred,color='red',label = "Regression Line")
plt.xlabel("MedInc")
plt.ylabel("MedHouseVal")
plt.title("Linear Regression")
plt.legend()
plt.grid(True)
plt.show()


# ----------------------Residual Graph---------------
residuals = y_test - y_pred
plt.scatter(y_pred,residuals,color='blue')
plt.axhline(y=0,color='red',linestyle='--')
plt.xlabel('predicated values')
plt.ylabel('Residuals')
plt.title('Residual Plot')
plt.legend()
plt.grid(True)
plt.show()
```
